```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import xgboost as xgb
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Device configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Define Bi-LSTM Model
class BiLSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, dropout_rate=0.3):
        super(BiLSTMModel, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers,
                            batch_first=True, bidirectional=True, dropout=dropout_rate)
        self.fc = nn.Linear(hidden_size * 2, 1)
        self.dropout = nn.Dropout(dropout_rate)  # Add dropout after LSTM

    def forward(self, x):
        h0 = torch.zeros(self.num_layers * 2, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers * 2, x.size(0), self.hidden_size).to(x.device)
        out, _ = self.lstm(x, (h0, c0))
        out = self.dropout(out[:, -1, :])  # Apply dropout before final FC layer
        return self.fc(out)


# Set Parameters
input_size = 3
hidden_size = 64
num_layers_list = [2, 3, 5]  # Different LSTM layers
learning_rate = 0.001
num_epochs = 100

# Convert data to PyTorch tensors
X_train_torch = torch.tensor(X_train.values, dtype=torch.float32).unsqueeze(1).to(device)
Y_train_torch = torch.tensor(Y_train.values, dtype=torch.float32).unsqueeze(1).to(device)
X_val_torch = torch.tensor(X_val.values, dtype=torch.float32).unsqueeze(1).to(device)
Y_val_torch = torch.tensor(Y_val.values, dtype=torch.float32).unsqueeze(1).to(device)
X_test_torch = torch.tensor(X_test.values, dtype=torch.float32).unsqueeze(1).to(device)
Y_test_torch = torch.tensor(Y_test.values, dtype=torch.float32).unsqueeze(1).to(device)



# Dictionary to store predictions
train_preds, val_preds, test_preds = [], [], []

# Train multiple Bi-LSTM models
for num_layers in num_layers_list:
    print(f"\nTraining Bi-LSTM with {num_layers} layers...")

    # Initialize model, loss function, and optimizer
    model = BiLSTMModel(input_size, hidden_size, num_layers).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    # Training loop
    for epoch in range(num_epochs):
        model.train()
        outputs = model(X_train_torch)
        loss = criterion(outputs, Y_train_torch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if (epoch + 1) % 10 == 0:
            print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}")

    # Evaluation
    model.eval()
    with torch.no_grad():
        train_pred = model(X_train_torch).cpu().numpy()
        val_pred = model(X_val_torch).cpu().numpy()
        test_pred = model(X_test_torch).cpu().numpy()

    # Store predictions for stacking
    train_preds.append(train_pred)
    val_preds.append(val_pred)
    test_preds.append(test_pred)

# Convert lists to numpy arrays for XGBoost training
X_train_stacked = np.hstack(train_preds)  # Stack predictions as features
X_val_stacked = np.hstack(val_preds)
X_test_stacked = np.hstack(test_preds)

# Convert target values to NumPy
Y_train_np = Y_train_torch.cpu().numpy().squeeze()
Y_val_np = Y_val_torch.cpu().numpy().squeeze()
Y_test_np = Y_test_torch.cpu().numpy().squeeze()


# Train XGBoost model on stacked features
xgb_model = xgb.XGBRegressor(objective="reg:squarederror", n_estimators=100, learning_rate=0.1)
xgb_model.fit(X_train_stacked, Y_train_np.ravel())

# Predictions from XGBoost
final_train_pred = xgb_model.predict(X_train_stacked)
final_val_pred = xgb_model.predict(X_val_stacked)
final_test_pred = xgb_model.predict(X_test_stacked)

# Compute metrics safely
def compute_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    # Handle small Y_true values
    eps = 1e-8  # Small constant to avoid division by zero
    mape = np.mean(np.abs((y_true - y_pred) / (y_true + eps))) * 100  # Adjusted MAPE

    return mae, mse, rmse, r2, mape



# Calculate metrics
metrics_train = compute_metrics(Y_train_np, final_train_pred)
metrics_val = compute_metrics(Y_val_np, final_val_pred)
metrics_test = compute_metrics(Y_test_np, final_test_pred)


# Print final metrics
print("\nFinal Stacked Ensemble Model Metrics:")
print("Training: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}, MAPE={:.2f}%".format(*metrics_train))
print("Validation: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}, MAPE={:.2f}%".format(*metrics_val))
print("Test: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}, MAPE={:.2f}%".format(*metrics_test))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import lightgbm as lgb
import ConfigSpace as CS
import ConfigSpace.hyperparameters as CSH
import hpbandster.core.nameserver as hpns
import hpbandster.core.result as hpres
from hpbandster.optimizers import BOHB
from hpbandster.core.worker import Worker
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import time

# Device configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Define Bi-LSTM Model
class BiLSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, dropout_rate):
        super(BiLSTMModel, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers,
                            batch_first=True, bidirectional=True, dropout=dropout_rate)
        self.fc = nn.Linear(hidden_size * 2, 1)
        self.dropout = nn.Dropout(dropout_rate)

    def forward(self, x):
        h0 = torch.zeros(self.num_layers * 2, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers * 2, x.size(0), self.hidden_size).to(x.device)
        out, _ = self.lstm(x, (h0, c0))
        out = self.dropout(out[:, -1, :])
        return self.fc(out)

# Best layer count determined earlier
BEST_LAYERS = 3

# Define Worker for BOHB
class MyWorker(Worker):
    def compute(self, config, budget, *args, **kwargs):
        hidden_size = config['hidden_size']
        dropout_rate = config['dropout_rate']
        learning_rate = config['learning_rate']

        model = BiLSTMModel(input_size=3, hidden_size=hidden_size, num_layers=BEST_LAYERS, dropout_rate=dropout_rate).to(device)
        criterion = nn.MSELoss()
        optimizer = optim.Adam(model.parameters(), lr=learning_rate)

        num_epochs = int(budget)  # Budget defines training epochs
        best_val_loss = float("inf")

        # Training with lower epochs for fast convergence
        for epoch in range(num_epochs):
            model.train()
            outputs = model(X_train_torch)
            loss = criterion(outputs, Y_train_torch)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            model.eval()
            with torch.no_grad():
                val_outputs = model(X_val_torch)
                val_loss = criterion(val_outputs, Y_val_torch)

            # Save best validation loss
            if val_loss.item() < best_val_loss:
                best_val_loss = val_loss.item()

        return {'loss': best_val_loss}

# Define the configuration space for BOHB
def get_config_space():
    config_space = CS.ConfigurationSpace()
    config_space.add_hyperparameter(CSH.UniformIntegerHyperparameter('hidden_size', lower=32, upper=96))  # Reduced upper limit
    config_space.add_hyperparameter(CSH.UniformFloatHyperparameter('dropout_rate', lower=0.1, upper=0.4))  # Lower dropout
    config_space.add_hyperparameter(CSH.UniformFloatHyperparameter('learning_rate', lower=1e-4, upper=5e-3, log=True))  # Narrow search
    return config_space

# Run BOHB optimization
start_time = time.time()

NS = hpns.NameServer(run_id='bohb_run', host='localhost', port=None)
NS.start()

worker = MyWorker(run_id='bohb_run', nameserver='localhost')
worker.run(background=True)

bohb = BOHB(configspace=get_config_space(), run_id='bohb_run', nameserver='localhost', min_budget=5, max_budget=30)  # Reduced budget
res = bohb.run(n_iterations=10)  # Lower iterations for faster tuning

# Get the best configuration
bohb.shutdown()
NS.shutdown()

# Get the best configuration
incumbent = res.get_incumbent_id()
id2config = res.get_id2config_mapping()
best_config = id2config[incumbent]['config']

print(f"Best Hyperparameters for {BEST_LAYERS} layers: {best_config}")
print(f"Tuning Completed in {time.time() - start_time:.2f} seconds")


# Train the best Bi-LSTM Model with optimized hyperparameters
best_model = BiLSTMModel(input_size=3, hidden_size=best_config['hidden_size'], num_layers=BEST_LAYERS, dropout_rate=best_config['dropout_rate']).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(best_model.parameters(), lr=best_config['learning_rate'])
num_epochs = 30  # Reduced epochs for faster training

for epoch in range(num_epochs):
    best_model.train()
    outputs = best_model(X_train_torch)
    loss = criterion(outputs, Y_train_torch)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Generate stacked features for LightGBM
best_model.eval()
with torch.no_grad():
    train_pred = best_model(X_train_torch).cpu().numpy()
    val_pred = best_model(X_val_torch).cpu().numpy()
    test_pred = best_model(X_test_torch).cpu().numpy()

# Train LightGBM model
lgb_train = lgb.Dataset(train_pred, label=Y_train_np)
lgb_val = lgb.Dataset(val_pred, label=Y_val_np, reference=lgb_train)

params = {
    'objective': 'regression',
    'metric': 'rmse',
    'learning_rate': 0.05,
    'num_leaves': 31,
    'verbose': -1
}

lgb_model = lgb.train(
    params,
    lgb_train,
    num_boost_round=50,  # Reduced rounds
    valid_sets=[lgb_train, lgb_val],
    callbacks=[lgb.early_stopping(stopping_rounds=5)]  # Faster stopping
)

# Predictions from LightGBM
final_train_pred = lgb_model.predict(train_pred)
final_val_pred = lgb_model.predict(val_pred)
final_test_pred = lgb_model.predict(test_pred)

# Compute evaluation metrics
def compute_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return mae, mse, rmse, r2

metrics_train = compute_metrics(Y_train_np, final_train_pred)
metrics_val = compute_metrics(Y_val_np, final_val_pred)
metrics_test = compute_metrics(Y_test_np, final_test_pred)

# Print final results
print("\nFinal Stacked Model Metrics:")
print("Training: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}".format(*metrics_train))
print("Validation: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}".format(*metrics_val))
print("Test: MAE={:.4f}, MSE={:.4f}, RMSE={:.4f}, R²={:.4f}".format(*metrics_test))
```

```python
import requests

BASE_URL = "https://www.data.gov.in/api/resource_download"
RESOURCE_ID = 86943  # Today's resource ID
HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

def fetch_csv(resource_id):
    url = f"{BASE_URL}?resource_id={resource_id}&file_type=csv"
    response = requests.get(url, headers=HEADERS)
    if response.status_code == 200 and 'csv' in response.headers.get('Content-Type', ''):
        print(f"Success: Found CSV for resource_id={resource_id}")
        return url  # Returning the valid URL
    return None

# Test a range of past resource IDs
days_back = 5  # Adjust as needed
for i in range(days_back):
    test_id = RESOURCE_ID - i  # Assuming sequential resource IDs
    csv_url = fetch_csv(test_id)
    if csv_url:
        print(f"Download CSV: {csv_url}")
        break  # Stop when we find a valid past CSV
```
